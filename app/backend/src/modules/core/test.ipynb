```python
from entity_dxf import EntityDxf

file = EntityDxf('teste.dxf')
```

```python
file.get_types_in_layers(None)
```

```python
for i in file.psp:
    # Verifica o tipo da entidade para saber se ela pode ter texto
    tipo = i.dxftype()
    
    if tipo == 'TEXT':
        # Texto de linha única usa o atributo .dxf.text
        conteudo = i.dxf.text
        print(f"TEXTO: {conteudo}")
        
    elif tipo == 'MTEXT':
        # MText (Multi-line) também usa .dxf.text, mas pode conter
        # códigos de formatação (ex: \P para nova linha)
        conteudo = i.dxf.text
        print(f"MTEXTO: {conteudo}")
        
    elif tipo == 'ATTRIB' or tipo == 'ATTDEF':
        # Se forem atributos de blocos (como em carimbos/selos)
        conteudo = i.dxf.text
        print(f"ATRIBUTO: {conteudo}")
```

```python
from collections import defaultdict

result = defaultdict(lambda: defaultdict(int))

for e in file.msp:
    layer = e.dxf.layer
    tipo = e.dxftype()
    result[layer][tipo] += 1


result
```

```python
import math
def calcula():
    soma = 0
    file.msp.query('LINE[layer=="ELE-FIOS"]')
    for e in file.msp:
        if e.dxftype() == 'LINE':
            start = e.dxf.start
            end = e.dxf.end
            comprimento = math.dist((start.x, start.y), (end.x, end.y))
            soma += comprimento
```
